```python
# Step 1: Imports
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder

# Step 2: Load and Prepare Data
df = pd.read_csv("data_bodmas.csv")

categorical_cols = ['Donor_BloodType', 'Recipient_BloodType', 'Donor_Rh', 'Recipient_Rh', 'Organ_Type']
label_encoders = {}

# Encode categorical features
for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    label_encoders[col] = le

# Step 3: Train Model
X = df.drop(columns=['Donor_ID', 'Recipient_ID', 'Is_Match'])
y = df['Is_Match']

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X, y)

# Step 4: Define Prediction Function
def predict_match(new_data: dict):
    input_df = pd.DataFrame([new_data])
    
    for col in categorical_cols:
        le = label_encoders[col]
        input_df[col] = le.transform([input_df[col][0]])
    
    prediction = model.predict(input_df)[0]
    return "Match" if prediction == 1 else "No Match"

# Step 5: Sample Input
sample_input = {
    "Donor_BloodType": "A",
    "Recipient_BloodType": "B",
    "Donor_Rh": "+",
    "Recipient_Rh": "-",
    "Organ_Type": "Kidney",
    "Donor_Age": 35,
    "Recipient_Age": 45,
    "Donor_Lat": 25.3,
    "Donor_Lon": 80.1,
    "Recipient_Lat": 25.5,
    "Recipient_Lon": 80.0
}

# Step 6: Run Prediction
result = predict_match(sample_input)
print("Prediction Result:", result)
```
